```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import scipy

print("Particle Physics Research Environment: READY")
print("NumPy:", np.__version__)
print("Pandas:", pd.__version__)
print("SciPy:", scipy.__version__)
```

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.title("First Scientific Plot")
plt.show()
```
